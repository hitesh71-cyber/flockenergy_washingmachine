# Task 2: Exploratory Analysis of Washing-Machine Use

**Objective**: Analyze washing machine appliance readings from the official cleaned REFIT dataset across 4 households at 1-minute resolution (Houses 1, 2, 3, and 5).

### Pipeline Steps:
1. Resample official cleaned REFIT readings to 1-minute mean.
2. Apply a plain-language cycle detection rule:
   - Active power threshold $\ge 20\text{ W}$.
   - Inter-phase idle tolerance $\le 5\text{ minutes}$ (bridges drum agitation and rinse pauses).
   - Minimum cycle duration $\ge 15\text{ minutes}$.
   - Peak power $\ge 100\text{ W}$ (filters standby noise).
3. Summarize data quality (availability, zero/standby time, cycle counts).
4. Analyze diurnal (hour of day) and weekly (day of week) usage profiles.
5. Produce the 3 required figures: 24h trace, cycle zoom-in, and cross-house comparison.
